```python
import pandas as pd
import numpy as np
import plotly.express as px

np.random.seed(42)

n = 1000

cities = ["Bengaluru", "Hyderabad", "Chennai", "Mumbai"]
categories = ["Electronics", "Furniture", "Clothing", "Groceries"]

df = pd.DataFrame({
    "OrderID": [f"ORD{i}" for i in range(1001, 1001 + n)],
    "OrderDate": pd.to_datetime(
        np.random.choice(
            pd.date_range("2026-01-01", "2026-09-30"),
            n
        )
    ),
    "City": np.random.choice(cities, n),
    "Category": np.random.choice(categories, n),
    "Quantity": np.random.randint(1, 11, n),
    "DiscountPct": np.random.choice(
        [0, 5, 10, 15, 20, 25],
        n
    )
})

prices = {
    "Electronics": 5000,
    "Furniture": 8000,
    "Clothing": 1200,
    "Groceries": 500
}

cost_rates = {
    "Electronics": 0.72,
    "Furniture": 0.68,
    "Clothing": 0.55,
    "Groceries": 0.82
}

df["Product"] = df["Category"].map({
    "Electronics": "Headphones",
    "Furniture": "Office Chair",
    "Clothing": "Jacket",
    "Groceries": "Food Pack"
})

df["UnitPrice"] = df["Category"].map(prices)

df["Revenue"] = (
    df["Quantity"] * df["UnitPrice"]
    * (1 - df["DiscountPct"] / 100)
).round(2)

df["Cost"] = (
    df["Quantity"] * df["UnitPrice"]
    * df["Category"].map(cost_rates)
).round(2)

df["Profit"] = (df["Revenue"] - df["Cost"]).round(2)

print(df.head())
print("Rows:", len(df))
print("Missing values:", df.isna().sum().sum())
```

```python
# overall business KPI's
total_revenue = df["Revenue"].sum()
total_profit = df["Profit"].sum()
total_orders = df["OrderID"].nunique()
profit_margin = total_profit / total_revenue * 100
```

```python
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Total Profit: ₹{total_profit:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
# city level analysis
city_summary = (
    df.groupby("City", as_index=False)
      .agg(
          Revenue=("Revenue", "sum"),
          Profit=("Profit", "sum"),
          Orders=("OrderID", "nunique")
      )
)

city_summary["ProfitMarginPct"] = (
    city_summary["Profit"] / city_summary["Revenue"] * 100
)

print(city_summary.round(2))
```

```python
sales_by_city_category = (
    df.groupby(["City", "Category"], as_index=False)
      .agg(Revenue=("Revenue", "sum"))
)

fig1 = px.bar(
    sales_by_city_category,
    x="City",
    y="Revenue",
    color="Category",
    barmode="group",
    title="Revenue by City and Product Category",
    labels={
        "Revenue": "Revenue (₹)",
        "City": "City",
        "Category": "Product Category"
    },
    hover_data={"Revenue": ":,.2f"}
)

fig1.update_layout(
    template="plotly_white",
    yaxis_tickprefix="₹",
    legend_title="Category"
)

fig1.show()
```

```python
df["Month"] = df["OrderDate"].dt.to_period("M").astype(str)

monthly_sales = (
    df.groupby(["Month", "Category"], as_index=False)
      .agg(Revenue=("Revenue", "sum"))
      .sort_values("Month")
)

fig2 = px.line(
    monthly_sales,
    x="Month",
    y="Revenue",
    color="Category",
    markers=True,
    title="Monthly Revenue Trend by Category",
    labels={
        "Month": "Month",
        "Revenue": "Revenue (₹)",
        "Category": "Product Category"
    },
    hover_data={"Revenue": ":,.2f"}
)

fig2.update_layout(
    template="plotly_white",
    yaxis_tickprefix="₹",
    hovermode="x unified"
)

fig2.show()
```

```python
fig3 = px.scatter(
    df,
    x="Revenue",
    y="Profit",
    color="Category",
    size="Quantity",
    hover_name="OrderID",
    hover_data=[
        "City",
        "Product",
        "DiscountPct",
        "Cost",
        "Quantity"
    ],
    title="Order-Level Revenue vs Profit",
    labels={
        "Revenue": "Revenue (₹)",
        "Profit": "Profit (₹)",
        "Category": "Product Category",
        "Quantity": "Quantity Sold"
    }
)

fig3.update_layout(
    template="plotly_white",
    xaxis_tickprefix="₹",
    yaxis_tickprefix="₹"
)

fig3.show()
```
