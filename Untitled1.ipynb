```python
import numpy as np
```

```python
sales = np.array([1200, 1500, 900, 2100, 1800])

print(sales)
```

```python
print("Dimension :", sales.ndim)
print("Shape     :", sales.shape)
print("Data type :", sales.dtype)
print("Total     :", sales.sum())
print("Agerage   :", sales.mean())
print("Maximum   :", sales.max())
```

```python
sales_matrix = np.array([
    [1200, 1500, 900],
    [2100, 1800, 1600],
    [900, 1100, 1400]
])

print("Sales Matrix :\n", sales_matrix)
```

```python
print('Shape :', sales_matrix.shape)
print("data :", sales_matrix[0, 2])
print("data :", sales_matrix[:, 0])
print("data :", sales_matrix[1, 1])
```

```python
prices = np.array([100, 250, 400, 150, 600])
quantities = np.array([10, 5, 3, 8, 2])

revenue = prices * quantities
print("revenue :", revenue)
print("Total revenue :", revenue.sum())
```

```python
prices = np.array([1000, 2000, 3000])
discount_rate = 0.10
discounted_price = prices * (1 - discount_rate)
print(discounted_price)
```

```python
# array Selection based on condition
sales = np.array([1200, 4500, 800, 6500, 3200, 900])
print("Sales :", sales)
high_sales = sales[sales > 3000]
print("High Sales :", high_sales)
```

```python
high_value = sales[(sales >= 2000) & (sales < 5000)] 
print(high_value)
```

```python
# aggregation with axis
data = np.array([
    [100, 200, 300],
    [400, 500, 600],
    [700, 800, 900]
])

print(data.sum(axis=0))       # sum each column
print(data.sum(axis=1))       # sum each row
```

```python
# weighed Averatge
prices = np.array([100, 200, 300])
qty = np.array([10, 20, 5])

weighed_price = np.average(prices, weights=qty)
print(weighed_price)
```

```python
# random sampling and reproducibility

rng = np.random.default_rng(42)
sample = rng.integers(1, 101, size=10)
print(sample)
```

```python
# linea algebra and numerical analysis
A = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([10, 15])

solution = np.linalg.solve(A, b)

print("Solution :", solution)


# A = [2, 1]
#     [1, 3]

#     coefficients 
#     2x + y = 10
#     x + 3y = 15

#     since  b = [10, 15]

#     x = 3 and y = 4
```
