```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 2000

df = pd.DataFrame({
    "CustomerID": [f"C{i:05d}" for i in range(1, n + 1)],
    "Tenure": rng.integers(1, 73, n),
    "MonthlyCharges": rng.integers(300, 3001, n),
    "Contract": rng.choice(
        ["Monthly", "Annual", "TwoYear"],
        n,
        p=[0.55, 0.30, 0.15]
    ),
    "SupportCalls": rng.poisson(2, n),
    "LatePayments": rng.poisson(1, n),
    "Satisfaction": rng.integers(1, 6, n)
})

# Create a synthetic churn probability.
# These rules are illustrative, not real-world estimates.
logit = (
    -1.5
    + 0.035 * (df["MonthlyCharges"] / 100)
    - 0.035 * df["Tenure"]
    + 0.45 * df["SupportCalls"]
    + 0.55 * df["LatePayments"]
    - 0.65 * (df["Satisfaction"] - 3)
    - 1.0 * (df["Contract"] == "Annual")
    - 1.6 * (df["Contract"] == "TwoYear")
)

probability = 1 / (1 + np.exp(-logit))

df["Churn"] = np.where(
    rng.random(n) < probability,
    "Yes",
    "No"
)

df.to_csv("customer_churn.csv", index=False)

print(df.head())
print("\nChurn distribution:")
print(df["Churn"].value_counts())
print("\nDataset shape:", df.shape)
```

```python

import pandas as pd
from pycaret.classification import ClassificationExperiment

# 1. Load historical data
df = pd.read_csv("customer_churn.csv")
df = df.drop(columns=["CustomerID"])

# 2. Initialize experiment using PyCaret 3.x syntax
exp = ClassificationExperiment()

exp.setup(
    data=df,
    target="Churn",
    session_id=42,
    train_size=0.80,
    fold=5,
    verbose=False,
    html=False
)

# 3. Compare classification algorithms
best_model = exp.compare_models(
    sort="AUC",
    n_select=1
)

print("Best model selected:")
print(best_model)

# 4. Tune the selected model
tuned_model = exp.tune_model(
    best_model,
    optimize="AUC",
    n_iter=10
)

# 5. Evaluate predictions on holdout data
holdout_predictions = exp.predict_model(tuned_model)

print("\nHoldout predictions:")
print(holdout_predictions.head())

# 6. Finalize the model using all available historical data
final_model = exp.finalize_model(tuned_model)

# 7. Save the complete model pipeline
exp.save_model(final_model, "churn_model")

print("\nModel saved as churn_model.pkl")

```

```python

import pandas as pd
from pycaret.classification import load_model, predict_model

# Load the saved model pipeline
model = load_model("churn_model")

# New customers: do not include the Churn target column
new_customers = pd.DataFrame([
    {
        "Tenure": 3,
        "MonthlyCharges": 2800,
        "Contract": "Monthly",
        "SupportCalls": 7,
        "LatePayments": 4,
        "Satisfaction": 1
    },
    {
        "Tenure": 48,
        "MonthlyCharges": 900,
        "Contract": "Annual",
        "SupportCalls": 1,
        "LatePayments": 0,
        "Satisfaction": 5
    },
    {
        "Tenure": 12,
        "MonthlyCharges": 1600,
        "Contract": "Monthly",
        "SupportCalls": 3,
        "LatePayments": 1,
        "Satisfaction": 3
    }
])

# Predict label and score
result = predict_model(
    model,
    data=new_customers
)

# PyCaret 3.x normally returns prediction_label and prediction_score
print(result[
    ["prediction_label", "prediction_score"]
])

# Save the predictions for Power BI
result.to_csv("customer_churn_predictions.csv", index=False)

print("\nPredictions saved successfully.")
```
