## Data Cleaning Missing values, duplicate Rows and invalid recs

```python
df_clean = df.drop_duplicates().copy()

print("Before :", len(df))
print("After :", len(df_clean))
```

```python
print("Duplicated rows :", df_clean.duplicated().sum())
```

```python
print(df_clean.isna().sum())
```

```python
df_clean["Quantity"] = df_clean["Quantity"].fillna(df_clean["Quantity"].median())
```

```python
df_clean["Discount"] = df_clean["Discount"].fillna(0)
```

```python
df_clean['City'] = df_clean['City'].fillna("Unknown")
```

```python
print(df_clean.isna().sum())
```

```python
Invalid_qty = df_clean[df_clean['Quantity'] <= 0]
Invalid_prc = df_clean[df_clean['UnitPrice'] < 0]

print("Invalid Quantity :", len(Invalid_qty))
print("Invalid Price :", len(Invalid_prc))
```

##  Feature Engineering with numpy and pandas

```python
df_clean["GrossSales"] = (df_clean["Quantity"] * df_clean["UnitPrice"])
df_clean["DiscountAmount"] = (df_clean["GrossSales"] * df_clean["Discount"])
df_clean["NetRevenue"] = (df_clean["GrossSales"] - df_clean["DiscountAmount"])
df_clean["TotalCost"] = (df_clean["Quantity"] * df_clean["UnitCost"])
df_clean["Profit"] = (df_clean["NetRevenue"] - df_clean["TotalCost"] )
df_clean["ProfitMargin"] =  df_clean["NetRevenue"]!=0
```

```python
df_clean['LowMargin'] = np.where(df_clean["ProfitMargin"] < 10, "Review", "Acceptable")
```
